```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud
from collections import Counter
import nltk

from nltk.corpus import stopwords

def warn(*args, **kwargs):
    pass
    
import warnings
warnings.warn = warn
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/-H0BcPr512mhbTDDfkcTJA/comments-with-emotions.csv')
df.info()
```

```python
#Step 1: Remove duplicates
df = df.drop_duplicates()

#Step 2: Remove rows with missing values
df = df.dropna()

#Step 3: Drop unnecessary columns
df = df.drop(columns=['published_at','author'])

df
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

#WordCloud for 'comment' column
text = " ".join(comment for comment in df['comment'].astype(str))
wordcloud = WordCloud(width=800, height=400, background_color='snow').generate(text)

#Plot WordCloud
plt.figure(figsize=(10,5), facecolor='whitesmoke')
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title("WordCloud of Comments", fontsize=18, color="black", fontweight='bold')
plt.show()
```

```python
import seaborn as sns

emotion_columns = ['anger','fear','negative','positive','trust','anticipation','joy','surprise','sadness','disgust']

correlation_matrix = df[emotion_columns].corr()

plt.figure(figsize=(10,8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', square=True)
plt.title("Correlation Heatmap of Emotions", fontsize=18)
plt.show()
```

```python
#Barplot of top emotions
mean_emotions = df[emotion_columns].mean().sort_values(ascending=False)
plt.figure(figsize=(10,5))
sns.barplot(x=mean_emotions.index, y=mean_emotions.values)
plt.title("Average Emotion Scores", fontsize=18)
plt.xlabel("Emotions")
plt.ylabel("Average Score")
plt.xticks(rotation=45)
plt.show()
```

```python
#Top 10 most common words in comments
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

all_words = " ".join(df['comment'].astype(str)).split()
filtered_words = [word.lower() for word in all_words if word.isalpha() and word.lower() not in stop_words]
word_counts = Counter(filtered_words)
most_common_words = word_counts.most_common(10)
words, counts = zip(*most_common_words)

plt.figure(figsize=(10,5))
sns.barplot(x=list(words), y=list(counts), palette='viridis')
plt.title("Top 10 Most Common Words in Comments", fontsize=18)
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.xticks(rotation=45)
plt.show()
```

```python
#Pairplot for selected emotion columns
selected_emotions = ['positive','negative','joy','sadness','anger']
sns.pairplot(df[selected_emotions], diag_kind='kde', corner=True)
plt.suptitle("Pairplot of Selected Emotion Scores", y=1.02, fontsize=18)
plt.show()
```

```python
#Average likes by sentiment
avg_likes = df.groupby('sentiment')['like_count'].mean()
plt.figure(figsize=(8,5))
avg_likes.plot(kind='bar', color='lightcoral')
plt.title("Average Likes by Sentiment", fontsize=18)
plt.xlabel("Sentiment")
plt.ylabel("Average Likes")
plt.xticks(rotation=40)
plt.show()
```

```python
from prettytable import PrettyTable

#Top 5 POsitive Emotion Comments
pd.set_option('display.max_colwidth', None)

#Comments with highest positive emotion scores
top_positive_comments = df.sort_values(by='positive', ascending=False).head(5)

#Correct syntax to print specific columns
# top_positive_comments[['comment', 'positive']]

table = PrettyTable()
table.field_names = ["Comment", "Positive Emotion Score"]

# Add row to the table
for _, row in top_positive_comments.iterrows():
    table.add_row([row['comment'], row['positive']])

table.align["Comment"] = "l"
table.hrules = True
table.max_width = 150

# Print the table
print(table)
```

```python
#Top 5 Negative Emotion Comments
top_positive_comments = df.sort_values(by='negative', ascending=False).head(5)

#Correct syntax to print specific columns
# top_positive_comments[['comment', 'negative']]
table = PrettyTable()
table.field_names = ["Comment", "Positive Emotion Score"]

# Add row to the table
for _, row in top_positive_comments.iterrows():
    table.add_row([row['comment'], row['negative']])

table.align["Comment"] = "l"
table.hrules = True
table.max_width = 150

# Print the table
print(table)
```
