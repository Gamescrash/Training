Pandas is a Python library that provides extensive means for data analysis. Data scientist often work with data stored in table formats like .csv , .xlsx. Pandas makes it very convenient to load, process and analyze such tabular data using SQL-like queries. In conjuction with $Matplotlib$ and $Seaborn$, Pandas provides a wide range of opportunities for visual analysis of tabular data.

```python
import numpy as np
import pandas as pd

pd.set_option("display.precision", 2)
```

```python
df = pd.read_csv("./dataset/telecom_churn.csv")
df.head()
```

```python
print(df.shape)
```

```python
print(df.columns)
```

```python
print(df.info())
```

```python
df["Churn"] = df["Churn"].astype("int64")
```

```python
df.describe()
```

```python
df.describe(include=["object", "bool"])
```

```python
df["Churn"].value_counts()
```

```python
df["Churn"].value_counts(normalize=True)
```

```python
df.sort_values(by="Total day charge", ascending=False).head()
```

```python
df.sort_values(by=["Churn", "Total day charge"], ascending=[True, False]).head()
```

```python
# Indexing and retireving data
df["Churn"].mean()
```

```python
# What are the average values of numeric columns for customers who churned?
df[df["Churn"]==1].mean(numeric_only=True)
```

```python
# How much time (on average) do churned customers users spend on the phone during daytime?
df[df["Churn"]==1]["Total day minutes"].mean()
```

```python
# What is the maximum lenght of international calls among loyal users (Churn=0) who do not have an international plan?
df[(df["Churn"]==0) & (df["International plan"]=="No")]["Total intl minutes"].max()
```

```python
#Applying Functions to Cells, Columns and Rows
df.apply(np.max)
```

```python
df[df["State"].apply(lambda state: state[0] == "W")].head()
```

```python
d = {"No": False, "Yes": True}
df["International plan"] = df["International plan"].map(d)
df.head()
```

```python
df = df.replace({"Voice mail plan": d})
df.head()
```

```python
# Grouping 
columns_to_show = ["Total day minutes", "Total eve minutes", "Total night minutes"]

df.groupby(["Churn"])[columns_to_show].describe(percentiles=[])
```

```python
df.groupby(["Churn"])[columns_to_show].agg([np.mean, np.std, np.min, np.max])
```
