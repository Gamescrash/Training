# Ensembles

In ensemble learning we train multiple models on the same dataset, then we do prediction on these multiple models and then whenever output, we combine that result somehow to get our final output

In ensembles we have two main techniques: Bagging and Boosting

---

```python
import pandas as pd

df = pd.read_csv("./ML_dataset/diabetes.csv")
df.head()
```

```python
df.isnull().sum() # Check for missing values
```

```python
df.describe() # Examine stat of dataset and look for outliers
```

```python
df.Outcome.value_counts() # Check for imbalance dataset
```

```python
X = df.drop('Outcome', axis='columns')
y = df.Outcome
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled[:3]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, stratify=y, random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score

scores = cross_val_score(DecisionTreeClassifier(), X, y, cv=5)
scores
```

```python
scores.mean()
```

```python
from sklearn.ensemble import BaggingClassifier

bag_model = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=100,
    max_samples=0.8,
    oob_score=True,
    random_state=42
)

bag_model.fit(X_train, y_train)
bag_model.oob_score_
```

```python
bag_model.score(X_test, y_test)
```

```python
# Now try again with cross validation

bag_model = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=100,
    max_samples=0.8,
    oob_score=True,
    random_state=42
)

scores = cross_val_score(bag_model, X, y, cv=5)
scores.mean()
```

```python
from sklearn.ensemble import RandomForestClassifier

scores = cross_val_score(RandomForestClassifier(), X, y, cv=5)
scores.mean()
```
