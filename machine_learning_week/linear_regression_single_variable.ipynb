```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv('./ML_dataset/homeprices.csv')
df
```

```python
%matplotlib inline
plt.xlabel('area(sqr ft)')
plt.ylabel('price(US$)')
plt.scatter(df.area, df.price, color='red', marker='+')
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['area']],df.price)
```

```python
reg.predict(np.array([[3300]]))
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
%matplotlib inline
plt.xlabel('area', fontsize=20)
plt.ylabel('price', fontsize=20)
plt.scatter(df.area, df.price, color='red', marker='+')
plt.plot(df.area, reg.predict(df[['area']]), color='blue')
```

```python
d = pd.read_csv('./ML_dataset/areas.csv')
d
```

```python
p = reg.predict(d)
```

```python
d['prices'] = p
d.to_csv("./ML_dataset/prediction.csv")
```

Predict the canada per capita income in the 2020

```python
df = pd.read_csv('./ML_dataset/canada_per_capita_income.csv')
df
```

```python
df.columns
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['year']], df['per capita income (US$)'])
```

```python
reg.predict(np.array([[2000]]))
```

```python
%matplotlib inline
plt.xlabel('year', fontsize=18)
plt.ylabel('capita income', fontsize=18)
plt.scatter(df.year, df['per capita income (US$)'], color='black', marker='*')
plt.plot(df.year, reg.predict(df[['year']]), color='blue')
```

```python
reg.predict(np.array([[2020]]))
```
